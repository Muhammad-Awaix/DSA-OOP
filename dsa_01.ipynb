## Day 01

```python
# for and while loop 
for i in range(5):
    print(i, end=" ")
print()
i=0
while i<5:
    print(i, end=" ")
    i+=1
```

```python
# square pattern
n = 5
for _ in range(n):
    print("*"*n)
```

```python
# hollow square pattern
n=5
for i in range(n):
    if i==0 or i==n-1:
        print("*"*n)
    else:
        print("*"+" "*(n-2)+"*")
```

```python
# triangle pattern
n=5
for i in range(n+1):
    print("*"*i)
```

```python
# reverse triangle pattern
n=5
for i in range(n, 0, -1):
    print("*"*i)
```

```python
# compound pattern
n=5
for i in range(n):
    print("*"*i)
for i in range(n, 0, -1):
    print("*"*i)
```

```python
# paramid pattern
n=5
for i in range(n):
    spaces = n-i-1
    stars = 2*i+1
    print(" "*spaces+"*"*stars)
```

```python
# inverted paramid pattern
n=5
for i in range(n, 0, -1):
    spaces = n-i
    stars = 2*i-1
    print(" "*spaces+"*"*stars)
```

```python
# Dimond pattern
n=5
for i in range(n-1):
    spaces = n-i-1
    stars = 2*i+1
    print(" "*spaces+"*"*stars)
for i in range(n, 0, -1):
    spaces = n-i
    stars = 2*i-1
    print(" "*spaces+"*"*stars)
```

```python
n=5
for i in range(n+1):
    print(str(i)*i)
```

```python
# floyd's triangle
n=5
num=1
for i in range(1, n+1):
    for _ in range(i):
        print(num, end = " ")
        num+=1
    print()
```

```python
# binary triangle
n = 5
for i in range(1, n + 1):
    for j in range(1, i + 1):
        print((i + j) % 2, end=" ")
    print()
```

## Day 02

```python
# remove duplicates
a = [12, 13, 15, 10, 0, -10, 11, 10, -10, 15]
b = []
for i in a:
    if i in b:
        continue
    else:
        b.append(i)
print(b)
```

```python
# largest number
a = [12, 13, 15, 10, 0, -10,50, 11, 10, -10, 15, 20]
# max(a)
b=a[0]
for i in a:
    if i>b:
        b=i
print(b)
```

```python
# min number
a = [12, 13, 15, 10, 0, -10,50, 11,-500, 10, -10, 15, 20, -100]
# min(a)
b=a[0]
for i in a:
    if i<b:
        b=i
print(b)
```
