```python
# What is self
class Animals():
    def __init__(self):
        self.voice = "Meow"
        self.legs = 4
        print(id(self)) 
    def instinct(self):
        print(f"Cat has {self.legs} legs and says {self.voice}")

# 1st Object
cat1 = Animals()
cat1.instinct()
print(id(cat1))
# Both has the same id

# 2nd object
cat2 = Animals()
cat2.instinct()
print(id(cat2))
```

```python
# megic methods
# __init__

class Fraction():
    def __init__(self, n, d):
        self.num = n
        self.denum = d
 # __str__ for printing objects
    def __str__(self):
        return f"{self.num}/{self.denum}"
# __add__ for adding the objects
    def __add__(self,other):
        temp_num = self.num*other.denum+other.num*self.denum
        temp_denum = self.denum*other.denum
        return f"{temp_num}/{temp_denum}"
# __sub__ for adding the objects
    def __sub__(self,other):
        temp_num = self.num*other.denum-other.num*self.denum
        temp_denum = self.denum*other.denum
        return f"{temp_num}/{temp_denum}"
# __mul__ for multiplying
    def __mul__(self,other):
        temp_num = self.num*other.num
        temp_denum = self.denum*other.denum
        return f"{temp_num}/{temp_denum}"
    


x = Fraction(3,4)
y = Fraction(5,5)
print(x+y)
print(x-y)
print(x*y)
```

```python
# Encapsulation hiding the instance variables and data using __ in front  
from datetime import date

class Human:
    def __init__(self):
        # instance variables 
        self.__name = input("Enter your name: ")
        self.__birth = int(input("Enter your birth year: "))
        self.age = date.today().year - self.__birth

    def greeting(self):
        return f"Welcome {self.__name} to the party"

    def allow(self):
        if self.age < 18:
            return "Access Denied"
        else:
            return "Access Approved"
        

    # now getter method for name just for examples 
    def get_name(self):
        return self.__name
    # setting method for name
    def set_name(self, new_name):
        if type(new_name) == str:
            self.__name = new_name
            print("Name changed")
        else:
            print("Not Allowed")


h1 = Human()
```
